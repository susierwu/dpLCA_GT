```python
import pickle

hydro_path = "hydro_v2_reservoir_dp_results_MY2030_2050.pkl"

with open(hydro_path, "rb") as f:
    hydro_dict = pickle.load(f)


hydro_dict
```

```python
def harmonize_country_keys(hydro_dict):
    country_map = {
         ", RoW,":", CN,",
         ", CA-QC,": ", US," 
    }

    new_dict = {}
    for k, v in hydro_dict.items():
        new_k = k
        for old, new in country_map.items():
            new_k = new_k.replace(old, new)
        new_dict[new_k] = v

    return new_dict
```

```python
hydro_dict_harmonized = harmonize_country_keys(hydro_dict)
hydro_dict_harmonized
```

```python
out_fname = "hydro_v2_reservoir_dp_region_harmonized.pkl"

with open(out_fname, "wb") as f:
    pickle.dump(hydro_dict_harmonized, f)

print(f"✓ Saved: {out_fname}")
```
