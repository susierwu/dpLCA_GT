```python
import os
import xarray as xr
import sys
import itertools
from itertools import zip_longest
from lca_algebraic import initProject
import pandas as pd
import numpy as np
import xarray as xr
import pickle
import matplotlib.pyplot as plt
import matplotlib
import matplotlib.patches as mpatches
plt.rcParams['font.family'] = 'Times New Roman'
plt.style.use('tableau-colorblind10') #seaborn-v0_8-whitegrid
```

## 1. get LCA results from each module 

```python
ds_v1 = xr.open_dataset('../pGWP_module_output/V1AB_init_BEVGT_pGWP100.nc')
ds_f2 = xr.open_dataset('../pGWP_module_output/F2_BEVGT_pGWP100.nc')
ds_v1b = xr.open_dataset('../pGWP_module_output/V1B_batrepl_BEVGT_pGWP100.nc') 
```

#### export all 25 excels to output

```python
v1ab_tech_list =  ['LFP', 'LSB', 'LTO', 'NCA', 'NMC622']
v1b_tech_list =   ['LFP', 'LSB', 'LTO', 'NCA', 'NMC622']



for b1 in v1ab_tech_list: 
    for b2 in v1b_tech_list: 
        df_list = []
    
        for sp in ds_v1.SSP.values:
            for vy in ds_v1.v_year.values: 
                IC =  ds_v1.Impact_category.values 
                vy_col = [vy] * len(ds_v1.v_year.values)
                sp_col = [sp] * len(ds_v1.SSP.values)
                
                v1_score = ds_v1.sel(tech_list = b1, SSP = sp, v_year = vy, v_module = 'V1AB_init')['value'].values
                f2_score = ds_f2.sel(tech_list = "grid", SSP = sp, v_year = vy, v_module = 'F2_fuel_elec')['value'].values
                v1b_score = ds_v1b.sel(tech_list = b2, SSP = sp, v_year = vy, v_module = 'V1B_repl')['value'].values
        
                df_colname = ["SSP", "v_year", "IC", "V1AB", "F2", "V1B" ]
                df = pd.DataFrame(zip(sp_col, vy_col, IC, v1_score, f2_score, v1b_score ), columns = df_colname)
                df_list.append(df)


        bev_df = pd.concat(df_list, ignore_index=True)
        #print(len(bev_df))

        f_suffix = str(b1) +  str(b2)
        f_name = "BEV_vlifecycle_" + f_suffix + "finalLCA.xlsx"
        bev_df.to_excel(os.path.join("output", f_name), index=False)
```

## 2. user input tech and plot

```python
v1ab_tech =  'NMC622' #['LFP', 'LSB', 'LTO', 'NCA', 'NMC622']
v1b_tech =  'NMC622' #['LFP', 'LSB', 'LTO', 'NCA', 'NMC622']
```

```python
df_list = []
for sp in ds_v1.SSP.values:
    for vy in ds_v1.v_year.values: 
        IC =  ds_v1.Impact_category.values 
        vy_col = [vy] * len(ds_v1.v_year.values)
        sp_col = [sp] * len(ds_v1.SSP.values)
        
        v1_score = ds_v1.sel(tech_list = v1ab_tech, SSP = sp, v_year = vy, v_module = 'V1AB_init')['value'].values
        f2_score = ds_f2.sel(tech_list = "grid", SSP = sp, v_year = vy, v_module = 'F2_fuel_elec')['value'].values
        v1b_score = ds_v1b.sel(tech_list = v1b_tech, SSP = sp, v_year = vy, v_module = 'V1B_repl')['value'].values

        df_colname = ["SSP", "v_year", "IC", "V1AB", "F2", "V1B" ]
        df = pd.DataFrame(zip(sp_col, vy_col, IC, v1_score, f2_score, v1b_score ), columns = df_colname)
        df_list.append(df)
```

```python
bev_df = pd.concat(df_list, ignore_index=True)
bev_df
#icev_df2 = icev_df[icev_df["IC"] != 'IPCC2021_default']
```

## 2.B plot 

```python
color_dict = {
    'F2':  '#1f77b4',   # blue
    'V1AB': '#ff7f0e',  # Orange
    'V1B': '0.63',   # Medium gray
}

fig, axs = plt.subplots(3, 3, figsize=(9, 9))

axs = axs.flatten()
for i in range(len(df_list)) :
    ax = axs[i]

    ax.ticklabel_format(axis='y', style='sci', scilimits=(0,0))
    axs[i].grid(True, linestyle='--', linewidth=0.5, color='gray')

    df = df_list[i]
    ddf = df[['IC', 'V1AB', 'F2', 'V1B']]    
    pivot_df = ddf.pivot_table(index='IC', columns='IC', values= [ 'V1AB', 'F2', 'V1B' ] ) 
    colors = [color_dict[col[0]] for col in pivot_df.columns]
    subtitle = str(df["SSP"][0]) + "-" + str(df["v_year"][0] )
    pivot_df.plot(kind='bar', ax=ax, stacked=True, legend=False, color=colors, title =subtitle, width=0.36, xlabel='' )
    axs[i].set_title(subtitle)

    # for ylables
    if i % 3 != 0:
        #axs[i].set_yticklabels([])  
        axs[i].set_ylabel('') 
    else: 
        axs[i].set_ylabel(r'GWP100 kg CO$_2$e')


    
    # same y-label for same SSP: 
    if i == 0  or i == 1 or i == 2 :   
        axs[i].set_ylim(0, 3.5e5)
        """ 
        # Set the y-axis ticks to integers (1, 2, 3), not working as tick showing as 1e5, 2e5
        import matplotlib.ticker as ticker
        def format_yticks(value, pos):
            if value >= 1e5:
                return f'{int(value/1e5)}e5'
            else:
                return int(value)
                
        yticks = [1e5, 2e5, 3e5]
        ax.set_yticks(yticks)
        # Apply scientific notation for y-axis labels
        ax.ticklabel_format(axis='y', style='sci', scilimits=(0, 0))
        # Create a ticker object with the custom formatter
        formatter = ticker.FuncFormatter(format_yticks)
        ax.yaxis.set_major_formatter(formatter)
        """ 
    elif i == 3  or i == 4 or i == 5 :   
        axs[i].set_ylim(0, 6.5e5 )
    else  :   
        axs[i].set_ylim(0, 7.5e5 )
    
    # for xlables    
    if i<6:
        axs[i].set_xticklabels([])  
        axs[i].set_xlabel('') 
    else:
        ax.set_xticklabels(ax.get_xticklabels(), rotation=60, ha='right')


# Add labels and title
fig.suptitle('BEV-GT, with NMC622 for both initial and battery replacement, vehicle lifecycle impacts [20years, 36,000km per year] ')
handles = [mpatches.Patch(color=color_dict['F2'], label='F2:grid_elec'),
           mpatches.Patch(color=color_dict['V1AB'], label='V1AB:init_veh_bat_manuf'),
           mpatches.Patch(color=color_dict['V1B'], label='V1B:bat_repl')]
fig.legend(handles=handles, loc='lower right', bbox_to_anchor=(0.98, -0.01), fancybox=False, ncol=2)

plt.xlabel("")
fig.supxlabel('Impact Category')
plt.tight_layout()
plt.show()
```

## 3. save all tech to pickle
each list has 
- value[0] as the bat_name:  {'V1AB_init': 'NMC622', 'V1B': 'NMC622'}
- value[1] is the dataframe for this bat_name

```python
all_tech_toget_formod = {'V1AB_init': ['LFP', 'LSB', 'LTO', 'NCA', 'NMC622'],
                         'V1B': ['LFP', 'LSB', 'LTO', 'NCA', 'NMC622'],
                     }
combinations = list(itertools.product(*all_tech_toget_formod.values()))
each_tech_toget_formod = [{key: value for key, value in zip(all_tech_toget_formod.keys(), combination)} for combination in combinations]
print(len(each_tech_toget_formod))
each_tech_toget_formod
```

```python
final_alltech_list = []
for v1ab in ['LFP', 'LSB', 'LTO', 'NCA', 'NMC622']:  
    for v1b in ['LFP', 'LSB', 'LTO', 'NCA', 'NMC622']:

        df_list = []
        for sp in ds_v1.SSP.values:
            for vy in ds_v1.v_year.values: 
                IC =  ds_v1.Impact_category.values 
                vy_col = [vy] * len(ds_v1.v_year.values)
                sp_col = [sp] * len(ds_v1.SSP.values)
                
                v1_score = ds_v1.sel(tech_list = v1ab, SSP = sp, v_year = vy, v_module = 'V1AB_init')['value'].values
                f2_score = ds_f2.sel(tech_list = "grid", SSP = sp, v_year = vy, v_module = 'F2_fuel_elec')['value'].values
                v1b_score = ds_v1b.sel(tech_list = v1b, SSP = sp, v_year = vy, v_module = 'V1B_repl')['value'].values
        
                df_colname = ["SSP", "v_year", "IC", "V1AB", "F2", "V1B" ]
                df = pd.DataFrame(zip(sp_col, vy_col, IC, v1_score, f2_score, v1b_score ), columns = df_colname)
                df_list.append(df)

        bev_df = pd.concat(df_list, ignore_index=True)
        final_alltech_list.append(bev_df)
```

```python
print(len(each_tech_toget_formod), len(final_alltech_list))
```

```python
combined_ds_techdict = list(zip(each_tech_toget_formod, final_alltech_list))
```
