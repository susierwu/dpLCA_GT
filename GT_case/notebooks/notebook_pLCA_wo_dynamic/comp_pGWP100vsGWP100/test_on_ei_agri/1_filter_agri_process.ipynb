```python
import pickle
import bw2io, bw2data
import wurst as ws
import wurst.transformations as wt
```

```python
# Step 1: Set path to the .pkl file
pkl_file = "ecoinvent 3.9.1.pkl"
# Step 2: Load the file
with open(pkl_file, "rb") as f:
    data = pickle.load(f)
# Step 3: Inspect the structure
print("Type of data:", type(data))
print(len(data))
```

```python
def exclude(func):
    """Negate a wurst filter."""
    return lambda x: not func(x)

def has_isic_starting_with_01():
    return lambda x: any(
        c[0] == "ISIC rev.4 ecoinvent" and c[1].startswith("01")
        for c in x.get("classifications", [])
    )


agri_filter = [
    has_isic_starting_with_01(), 
    ws.equals("unit", "kilogram"),
    ws.equals("location", "RoW"),
    exclude(ws.contains("name", "market for")),  # ✅ negation using your exclude()
]

"""
agri_db = wurst.get_many(
    data, 
    *agri_filter
)

for ds in agri_db:
    pass
    print(ds["name"] )
"""
```

```python
def filter_agri_db(data, agri_filter):
    ag_db = []
    agri_db = ws.get_many(
        data, 
        *agri_filter
        )

    for ds in agri_db:
        try:
            wt.relink_technosphere_exchanges(
                ds,
                data
            )
        except:
            pass
            
        print(ds['name'], "|", ds['location'],  "|", ds['reference product'])
        ag_db.append(ds)

    return ag_db

    
```

```python
aad = filter_agri_db(data, agri_filter)
```
