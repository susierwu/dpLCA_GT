```python
#import bw2data, bw2io
#import bw2calc
import os
import matplotlib
import matplotlib.pyplot as plt
from matplotlib import colormaps
import numpy as np
import pandas as pd
import xarray as xr
plt.rcParams['font.family'] = 'Times New Roman'
matplotlib.style.use('tableau-colorblind10') #seaborn-v0_8-whitegrid  tableau-colorblind10
```

### prepare dIRF  under SSPx, MY2020/2030/2040/2050

```python
len_cf = 101  # add 1 as IRF starting from year_0
MYs = [2020, 2030, 2040, 2050] #, 2060, 2070]
ssps = ["119", "245", "585"]
sp_col, my_col = [], []

for sp in ssps: 
    repeat_sp = [sp] * len_cf * len(MYs)
    sp_col.append(repeat_sp)
sp_col_f = my_col_f = [item for sublist in sp_col for item in sublist]

for my in MYs: 
    repeat_my = [my] * len_cf  
    my_col.append(repeat_my)

my_col = my_col * len(ssps)
my_col_f = [item for sublist in my_col for item in sublist]

print(len(sp_col_f), len(my_col_f))
```

```python
year_col =  [np.arange(len_cf)] *len(ssps) * len(MYs)
year_col_f =  [item for sublist in year_col for item in sublist]
print(len(year_col_f))
```

```python
ghggass = ["CO2", "CH4", "N2O"]
cf_col_gas_allgas = []

for ghggas in ghggass: 
    cf_col_gas = []
    for sp in ssps:
        for MY in MYs:
            #print(f"read point IRF value for {ghggas} SSP {sp}, model year {MY}")
            filename = "agwp_dcf_gwp100_tstep1" + ghggas + "_ssp" + sp + "_fair_start2000MY" + str(MY) + ".xlsx"
            cf_file = os.path.join("/Users/susierwu/Dropbox/DLCA_DynCO2/pdpLCA_GT/FaIR_dpCFs/output/metrics/", filename)
            cf_gwp_df = pd.read_excel(cf_file, sheet_name = "rf_pointvalue" + ghggas+ "_" + str(MY) ) 
            cf_gwp_pointvalue = cf_gwp_df[ghggas].values
            cf_col_gas.append(cf_gwp_pointvalue)
    cf_col_gas_allgas.append(cf_col_gas)
```

```python
cf_col_gas_f = [item for sublist in cf_col_gas for item in sublist]
print(len(cf_col_gas_f))
```

```python
#co2_irf_all, ch4_irf_all, n2o_irf_all = cf_col_gas_allgas[0], cf_col_gas_allgas[1], cf_col_gas_allgas[2] 
gas_all_f_list = []
# gas_irf_all_f_list[i] is the ith gas from ghggass, flattened array for all SSPs, MYt in order 
for i in range(len(cf_col_gas_allgas)): 
    gas_all_f = [item for sublist in cf_col_gas_allgas[i] for item in sublist]
    gas_all_f_list.append(gas_all_f)
```

```python
cf_df = pd.DataFrame(zip(sp_col_f, my_col_f, year_col_f, gas_all_f_list[0], gas_all_f_list[1], gas_all_f_list[2]) , 
                     columns = ["SSP", "ModelYear", "Year", "CO2_IRF", "CH4_IRF", "N2O_IRF" ] ) 
```

#### for plotting, change 119 to 1-19 ...

```python
cf_df['SSP'] = cf_df['SSP'].str.replace(r'^(\d)(\d{2})$', r'\1-\2', regex=True)
```

```python
cf_df.set_index(['SSP', 'ModelYear', 'Year'], inplace=True)
```

```python
# to see % diff by SSP[x]:
cf_df.to_excel("IRF_output_allyears_N1212.xlsx", index = False)
```

```python
# Deleting rows with YEAR_index value 0 to make same shape as GWP
cf_df_filtered = cf_df[~cf_df.index.get_level_values("Year").isin([0])]
cf_df_filtered
```

```python
cf_xr = cf_df.to_xarray() 
```

### plotting dIRF[0-100], not [1-100] as for (A)GWP

```python
fig, axs = plt.subplots(1, 3 , figsize=(9, 3))
cf_xr.sel(ModelYear = 2030)['CO2_IRF'].plot(x='Year', hue='SSP', ax=axs[0], add_legend = True)
cf_xr.sel(ModelYear = 2040)['CO2_IRF'].plot(x='Year', hue='SSP', ax=axs[1], add_legend = False)
cf_xr.sel(ModelYear = 2050)['CO2_IRF'].plot(x='Year', hue='SSP', ax=axs[2], add_legend = False)

axs[0].set_ylabel(r'IRF[CO$_2$]  W m$^{-2}$ kg$^{-1}$', fontsize=13)
axs[1].set_ylabel('')
axs[2].set_ylabel('')

axs[0].set_xlabel('')
axs[1].set_xlabel('')
axs[2].set_xlabel('')


for ax in axs:
    ax.set_ylim(0.6e-15, 2e-15)
    ax.set_yticks( np.linspace(0.6e-15, 2e-15, 5) )

    
#fig.suptitle(r'Dynamic and prospective IRF for CO$_2$ per ModelYear', fontsize=12)
fig.suptitle(r'CO$_2$', fontsize=14)
plt.tight_layout()
plt.savefig('output_plot/IRF_CO2.jpg', dpi=300)
plt.show()
```
